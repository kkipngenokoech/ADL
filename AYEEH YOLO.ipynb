# 3.) Train the model

```python
# Load a pre-trained YOLO model
model = YOLO("yolo11n.pt")


# Fine tune to dog breed dataset task

# Can change number of epochs and other parameters - see: https://docs.ultralytics.com/modes/train/
model.train(data="/kaggle/input/yolo-dataset/YoloDemo/dataset.yaml",epochs=35)
```

# 4. Check where model is saved and move to YoloDemo folder

```python
import shutil

best_model_path = '/kaggle/working/runs/detect/train/weights/best.pt'
destination_path = '/kaggle/working/runs'

shutil.move(best_model_path, destination_path)
```

# 5.) Inference with your model!

```python
model_path = '/kaggle/working/runs/best.pt'
dog_model = YOLO(model_path)
```

```python
# Test Image Paths - A few examples, or add your own!
CMU_test = '/kaggle/input/yolo-dataset/YoloDemo/CMU_Scottie.png'
BCB_test = '/kaggle/input/yolo-dataset/YoloDemo/Boxer_BC_test.png'
Boxer_test =  '/kaggle/input/yolo-dataset/YoloDemo/Boxer.png'
GR_Beagle_test =  '/kaggle/input/yolo-dataset/YoloDemo/GR_Beagle_test.png'
frenchies =  '/kaggle/input/yolo-dataset/YoloDemo/frenchies.png'
golden =  '/kaggle/input/yolo-dataset/YoloDemo/Golden.png'
```

```python
# Run inference on the image
results = dog_model.predict(golden)
result = results[0]
result.show()
```
